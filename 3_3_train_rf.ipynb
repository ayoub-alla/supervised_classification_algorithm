## 3_3 Training using Random Forest

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix , PrecisionRecallDisplay , balanced_accuracy_score, matthews_corrcoef
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score, KFold
import numpy as np
from sklearn.model_selection import RandomizedSearchCV , learning_curve
from sklearn.utils import resample
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('ds/Fraud_cleaned_dataset.csv' , nrows=300000)
data.head(1)
```

```python
# target
y = data['is_fraud']

# features rest columns
X = data.drop(columns=['is_fraud'])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
#search best parameters

param_distributions = {
    'max_features': ['sqrt' , 'log2'],
    'max_depth': range(10, 40 , 10)
}


random_search = RandomizedSearchCV(
    estimator = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42),
    param_distributions=param_distributions,
    n_iter=10,
    cv=3 , # 3k validation fold
    scoring='balanced_accuracy', #  2 class equaly
    random_state=42,
    n_jobs=-1 
)

#we take  100 000 as a sample for fast searching and to minimise time
X_sample = X_train.sample(100000, random_state=42)
y_sample = y_train.loc[X_sample.index]

random_search.fit(X_sample, y_sample)
```

```python
#params found by random search
print(f"parameters : {random_search.best_params_}")
print(f"  balanced_accuracy : {random_search.best_score_:.4f}")

#sqrt max_depth 10
```

```python
#random forest Model

random_forest_model = RandomForestClassifier(max_features='sqrt' , max_depth=10, class_weight='balanced', random_state=42)

random_forest_model.fit(X_train, y_train)

#predictions
y_pred_rf = random_forest_model.predict(X_test)
```

```python
#reports
print(classification_report(y_test, y_pred_rf))
```

```python
#validation croise k fold 4 splits

kf = KFold(n_splits=4, shuffle=True, random_state=42)

scores = cross_val_score(random_forest_model, X, y, cv=kf, scoring='precision_macro')

print('based on precision_macro (average of both class ) ')
print(f"cross validation 4 Parts : {scores}")
print(f" Average: {np.mean(scores):.2f}")
```

```python
#confusion matric
tn, fp, fn, tp = confusion_matrix(y_test, y_pred_rf).ravel()
print(f" TP: {tp} , TN: {tn} , FP: {fp} , FN: {fn} , ")
```

```python
#metrics avancee

bal_acc = balanced_accuracy_score(y_test, y_pred_rf)
mcc = matthews_corrcoef(y_test, y_pred_rf)

print("Balanced Accuracy: " , bal_acc)
print("MCC Score:        " , mcc)
```

```python
X_sample, y_sample = resample(X, y, n_samples=150_000, stratify=y, random_state=42)

train_sizes, train_scores, test_scores = learning_curve(
    random_forest_model, X_sample, y_sample, cv=3, scoring='balanced_accuracy',
    train_sizes=np.linspace(0.5, 1.0, 5),
    error_score=np.nan,
    n_jobs=-1,   
)

train_mean = train_scores.mean(axis=1)
test_mean = test_scores.mean(axis=1)

plt.plot(train_sizes, train_mean, label='Train')
plt.plot(train_sizes, test_mean, label='Validation')
plt.xlabel('Training Size')
plt.ylabel('Balanced Accuracy')
plt.legend()
plt.show()
```

```python
display = PrecisionRecallDisplay.from_estimator(
    random_search.best_estimator_, X_test, y_test, name="random forest"
)
plt.title("Precision-Recall Curve")
plt.show()
```
